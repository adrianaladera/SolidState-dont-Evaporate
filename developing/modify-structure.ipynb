If we want to move our selected species for the [X, Y, Z] directions, then for a given species, set its selective dynamics = [True, True, True]. If we want it to remain static during relaxation in all directions, then selective dynamics = [False, False, False].

```python
from rdkit import Chem
from rdkit.Chem import AllChem
from ase import Atoms
from ase.io import write
import numpy as np
import pubchempy as pcp

smiles_list = {"3-methoxybenzeneselenol": "COC1=CC(=CC=C1)[SeH]",
            #    "3-methoxybenzenetellurol": "",
                # "4-methoxybenzenethiol": ","
               "2,4-dimethylbenzenethiol" :"CC1=CC(=C(C=C1)S)C",
               "2,6-dimethylbenzeneselenol": "CC1=C(C(=CC=C1)C)[SeH]",
                "2,6-dimethylbenzenetellurol": "CC1=C(C(=CC=C1)C)[TeH]",
                " 2-methylbenzenethiol": "CC1=CC=CC=C1S",
                "3-methylbenzenethiol" : "CC1=CC(=CC=C1)S",
                "4-methylbenzenethiol" : "CC1=CC=C(C=C1)S",
                "phenanthrene-9-selenol" : "[SeH]c1c2ccccc2c2ccccc2c1",
                "pentane-2-thiol" : ""
                }

# smiles_keys = [
#     # "3-methoxybenzeneselenol",
#     "3-methoxybenzenetellurol",
#     "4-methoxybenzenethiol",
#     "2,4-dimethylbenzenethiol",
#     "2,6-dimethylbenzeneselenol",
#     "2,6-dimethylbenzenetellurol",
#     "2-methylbenzenethiol",
#     "3-methylbenzenethiol",
#     "4-methylbenzenethiol",
#     "phenanthrene-9-selenol",
#     "pentane-2-thiol",
#     "hexane-3-thiol",
#     "heptane-4-thiol",
#     "benzenetellurol",
#     "4-methoxybenzenetellurol"
# ]

smiles_keys = ["2-methoxybenzenethiol"]

import os
from rdkit import Chem
from rdkit.Chem import AllChem
from ase import Atoms
from ase.io import write
import numpy as np
import pubchempy as pcp

# Fallback SMILES for exotic compounds
fallback_smiles = {
    "3-methoxybenzenetellurol":"COC1=CC(=CC=C1)[TeH]",
    "4-methoxybenzenetellurol":"COC1=CC=C(C=C1)[TeH]",
}

def build_ligand(iupac_name, write=False):
    # compound_list = pcp.get_compounds(iupac_name, namespace='name')
    # print(compound_list[0].to_dict()["canonical_smiles"])
    props = pcp.get_properties('IsomericSMILES', iupac_name, 'name')
    # print(props[0]["SMILES"])
    
    if props:
        smiles = props[0]["SMILES"]
    elif iupac_name in fallback_smiles:
        smiles = fallback_smiles[iupac_name]
    else:
        print(f"{iupac_name} not found 😢")
        return

    print(f"{iupac_name}: {smiles}")

    # RDKit molecule generation
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        print(f"Could not parse SMILES for {iupac_name} 😢")
        return

    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.UFFOptimizeMolecule(mol)

    # Extract 3D coordinates
    conf = mol.GetConformer()
    positions = np.array([[conf.GetAtomPosition(i).x,
                           conf.GetAtomPosition(i).y,
                           conf.GetAtomPosition(i).z] 
                          for i in range(mol.GetNumAtoms())])
    symbols = [atom.GetSymbol() for atom in mol.GetAtoms()]

    atoms = Atoms(symbols=symbols, positions=positions)

    if write:
        # set PBC
        atoms.set_cell([15.0, 15.0, 15.0])
        atoms.center()
        atoms.set_pbc([True, True, True])

        root = "/Users/adrianaladera/Desktop/MIT/research/VASP_calculations/fluorescent_1D/ligands/"
        folder = f"{root}{iupac_name.replace(' ', '_')}"
        os.makedirs(folder, exist_ok=True)
        write(f"{folder}/POSCAR.vasp", atoms, format="vasp")

    return atoms



for s in smiles_keys:
    build_ligand(s)


```
